```python
# 練習：if 條件的應用，判斷正負數
## 輸入一個整數(正負皆可)，如果不是整數則無條件去掉小數
## 回覆：正整數 或 負整數
## 注意：
##  輸入不能空白
##  可以不斷的輸入，直到0代表結束

while True:
  n = int( float(  input("請輸入一個整數：").strip() ) )
  # n = input("請輸入一個整數：")
  # n = n.strip()
  # n = float(n)
  # n = int(n)
  if n==0:
    print("程式結束")
    break
  elif n>0:
    print(f"{n}是正整數")
  else:
    print(f"{n}是負整數")

```

```python
# 練習：if 條件的應用，判斷奇偶數
## 輸入一個整數(正負皆可)，如果不是整數則無條件去掉小數
## 回覆：奇數 或 偶數
## 注意：
##  輸入不能空白
##  可以不斷的輸入，直到0代表結束

while True:
  n = int( float(  input("請輸入一個整數：").strip() ) )

  if n==0:
    print("程式結束")
    break
  elif n % 2 == 0:  # 若n除以2的餘數為0，則為偶數
    print(f"{n}是偶數")
  else:             # 否則為奇數
    print(f"{n}是奇數")
```

```python
# 練習：if 條件的應用
## 輸入姓名：李大仁 和 性別 m(男)、f(女)
## 回覆：李大仁先生 或 李大仁小姐
##
## 注意：
##  姓名不能空白
##  性別只能 m 或 f (大小寫都可以)
##
## 參考函數：
## • lower(): 全部大寫
## • upper(): 全部大寫
## • capitalize()：只有整句的第一個字母大寫，其餘字母全部變小寫
## • title()：每個單字的開頭字母都大寫(標題格式)

while True:
  name = input("請輸入姓名：").strip()
  if name != "":
    break
  else:
    print("姓名不能空白")

while True:
  gender = input("請輸入性別(M/F)：").upper()
  if gender == "M" or gender == "F":
    break
  else:
    print("性別只能是 M 或 F")

if gender == "M":
  gender = "先生"
else:
  gender = "小姐"

print(f"{name}{gender}")
```

```python
## while 迴圈的應用
## 一直輸入數字直到輸入0或空白

while True:
  n = input("請輸入數字(留空或0結束)：").strip()

  if n == "" or n == "0":
    print("結束！")
    break

  print(f"你輸入的是{n}")
```
