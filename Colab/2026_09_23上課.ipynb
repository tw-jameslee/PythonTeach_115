```python
# 條件敘述
# if ... elif ... else

# input的結果會是字串，所以須要用int()轉換型別
age = int(  input("請輸入你的年紀：")  )

if age<18:
  print("你是未成年！")
  print("可以買半票")
else:
  print("你已成年！")
  print("需要買全票")
```

```python
# input的結果會是字串，所以須要用int()轉換型別
age = int(  input("請輸入你的年紀：")  )

if age<8:
  print("你是未成年！")
  print("可以免買票")

elif age<18:
  print("你是未成年！")
  print("可以買半票")

elif age>=65:
  print("你是老人！")
  print("可以買6折票")

else:
  print("你已成年！")
  print("需要買全票")
```

```python
# 迴圈：while ...
# 迴圈：for ...

x = int( input("請輸入一個正或負的整數：") )

while x != 0:

  if x>0:
    print(f"{x}是一個正數")
  else:
    print(f"{x}是一個負數")

  x = int( input("請輸入一個正或負的整數：") )

print("你輸入了0，程式結束")
```

```python
# 作業一：終極密碼

# 由電腦自動產生一個整(1~100)
# 請使用者去猜，然後由程式告訴使用者太大、太小、猜對了

import random

# 先由電腦產生答案ans
ans = random.randint(1, 100)
print(f"答案是: {ans}")

g = int( input("請猜一個數字(1~100)：") )
while g != ans:  # 如果猜的和答案不同
  if ans>g:
    print("大一點！")
  else:
    print("小一點！")

  g = int( input("請猜一個數字(1~100)：") )

print("恭喜你，猜對了！")
```
